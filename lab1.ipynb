```python
  from collections import deque

graph={
    'A':['B','C'],
    'B':['D','E'],
    'C':['F','G'],
    'D':['H'],
    'E':['I'],
    'F':['J'],
    'G':[],
    'H':[],
    'I':[],
    'J':[]

}

def bfs(graph, start, target):
  if start not in graph or target not in graph:
    return None

  queue = deque([start])
  visited = {start}
  parent = {start: None}

  while queue:
    node = queue.popleft()

    if node == target:

      path = []
      current = target
      while current is not None:
        path.append(current)
        current = parent[current]
      return path[::-1]

    for neighbor in graph[node]:
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)
        parent[neighbor] = node

  return None

if __name__ == "__main__":
    print("Breadth-First Search")
    user_start = input("Enter starting letter: ").strip().upper()
    user_goal = input("Enter target goal letter: ").strip().upper()

    result_path = bfs(graph, user_start, user_goal)

    if result_path:
        print(f"\nPath found from {user_start} to {user_goal}:")
        print(" -> ".join(result_path))
    else:
        print("\nNode not found!")
```
