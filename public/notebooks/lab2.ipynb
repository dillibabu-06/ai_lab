```python
import string
from queue import PriorityQueue
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G', 'H'],
    'E': ['I', 'J'],
    'F': [],
    'G': [],
    'H': [],
    'I': [],
    'J': []
}

def heuristic(node, goal):
    return abs(ord(goal) - ord(node))

def best_first_search(start_node, goal_node, graph):
    if start_node not in graph or goal_node not in graph:
        return None
    pq = PriorityQueue()
    pq.put((heuristic(start_node, goal_node), start_node, [start_node]))

    visited = set()

    while not pq.empty():
        _, current, path = pq.get()

        if current == goal_node:
            return path

        if current not in visited:
            visited.add(current)

            for neighbor in graph.get(current, []):
                if neighbor not in visited:
                    new_path = path + [neighbor]
                    pq.put((heuristic(neighbor, goal_node), neighbor, new_path))

    return None

if __name__ == "__main__":
    print("Best-First Search")
    user_start = input("Enter starting letter").strip().upper()
    user_goal = input("Enter target goal letter").strip().upper()
    result_path = best_first_search(user_start, user_goal, graph)

    if result_path:
        print(f"\nPath found from {user_start} to {user_goal}:")
        print(" -> ".join(result_path))
    else:
        print("\nNode not found!")
```
