```python
from collections import deque

graph={
    'A':['B','C'],
    'B':['D','E'],
    'C':['F','G'],
    'D':['H'],
    'E':['I'],
    'F':['J'],
    'G':[],
    'H':[],
    'I':[],
    'J':[]

}

def dfs(graph, start, target):
  if start not in graph or target not in graph:
    return None

  stack = [start]
  visited = {start}
  parent = {start: None}

  while stack:
    current_node = stack.pop()

    if current_node == target:

      path = []
      node = target
      while node is not None:
        path.append(node)
        node = parent[node]
      return path[::-1]
    for neighbor in reversed(graph[current_node]):
      if neighbor not in visited:
        visited.add(neighbor)
        stack.append(neighbor)
        parent[neighbor] = current_node

  return None

if __name__ == "__main__":
    print("Depth-First Search")
    user_start = input("Enter starting letter: ").strip().upper()
    user_goal = input("Enter target goal letter: ").strip().upper()

    result_path = dfs(graph, user_start, user_goal)

    if result_path:
        print(f"\nPath found from {user_start} to {user_goal}:")
        print(" -> ".join(result_path))
    else:
        print("\nNode not found!")
```
