```python
class Node:
    def __init__(self, name, heuristic=0):
        self.name = name
        self.heuristic = heuristic
        self.solved = False
        self.children = []


def ao_star(node):


    if node.name == "J":
        node.solved = True
        node.heuristic = 0
        return

    if not node.children:
        return

    best_cost = float('inf')
    best_group = None


    for group in node.children:
        cost = sum(edge_cost + child.heuristic
                   for child, edge_cost in group)

        if cost < best_cost:
            best_cost = cost
            best_group = group


    for child, edge_cost in best_group:
        ao_star(child)

    node.heuristic = sum(
        edge_cost + child.heuristic
        for child, edge_cost in best_group
    )


    if all(child.solved for child, edge_cost in best_group):
        node.solved = True


def print_solution(node, level=0):

    print("  " * level + node.name)

    if node.name == "J" or not node.children:
        return

    best_cost = float('inf')
    best_group = None

    for group in node.children:

        cost = sum(
            edge_cost + child.heuristic
            for child, edge_cost in group
        )

        if cost < best_cost:
            best_cost = cost
            best_group = group

    for child, edge_cost in best_group:
        print_solution(child, level + 1)


A = Node("A", 10)
B = Node("B", 9)
C = Node("C", 8)
D = Node("D", 7)
E = Node("E", 6)
F = Node("F", 5)
G = Node("G", 4)
H = Node("H", 3)
I = Node("I", 2)
J = Node("J", 0)



A.children = [
    [(B, 1)],
    [(C, 2)]
]

B.children = [
    [(D, 2)],
    [(E, 3)]
]

C.children = [
    [(F, 2)],
    [(G, 1)]
]

D.children = [
    [(H, 2)],
    [(I, 3)]
]

E.children = [
    [(J, 1)]
]



ao_star(A)

print("Best solution:")
print_solution(A)

print("Cost =", A.heuristic)
```
