```python
import heapq

graph = {
    'A': [('B', 1), ('C', 3)],
    'B': [('D', 3), ('E', 2)],
    'C': [('F', 4)],
    'D': [],
    'E': [('G', 1)],
    'F': [],
    'G': [('H', 1)],
    'H': [('I', 1)],
    'I': [('J', 1)],
    'J': []
}

heuristic = {
    'A': 7,
    'B': 6,
    'C': 7,
    'D': 3,
    'E': 4,
    'F': 3,
    'G': 3,
    'H': 2,
    'I': 1,
    'J': 0
}

def astar(start, goal):

    if start not in graph or start not in heuristic:
        print(f"Error: Start node '{start}' is not in the graph")
        return
    if goal not in graph or goal not in heuristic:
        print(f"Error: Goal node '{goal}' is not in the graph")
        return

    pq = [(heuristic[start], 0, start, [start])]
    visited = set()

    while pq:
        f, cost, node, path = heapq.heappop(pq)

        if node == goal:
            print("Shortest Path:")
            print(" -> ".join(path))
            print("Cost =", cost)
            return

        if node not in visited:
            visited.add(node)

            for neighbor, c in graph[node]:
                if neighbor not in visited:
                    heapq.heappush(
                        pq,
                        (cost + c + heuristic[neighbor],
                         cost + c,
                         neighbor,
                         path + [neighbor])
                    )

    print(f"No path found from {start} to {goal}.")

start_node_input = input("Enter Start Node: ").upper()
goal_node_input = input("Enter Goal Node: ").upper()

astar(start_node_input, goal_node_input)
```
