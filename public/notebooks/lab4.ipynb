```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H', 'I'],
    'E': ['J'],
    'F': [],
    'G': [],
    'H': [],
    'I': [],
    'J': []
}

heuristic = {
    'A': 3,
    'B': 2,
    'C': 10,
    'D': 5,
    'E': 1,
    'F': 11,
    'G': 12,
    'H': 6,
    'I': 7,
    'J': 0
}

def heuristic_search(start, goal):
    current = start
    path = [current]

    while current != goal:
        neighbors = graph[current]

        if not neighbors:
            print("Goal not found!")
            return


        current = min(neighbors, key=lambda x: heuristic[x])
        path.append(current)

    print("\nGoal Found!")
    print("Path:")
    print(" -> ".join(path))

start = input("Enter the starting node: ").upper()
goal = input("Enter the goal node: ").upper()

if start in graph and goal in graph:
    heuristic_search(start, goal)
else:
    print("Invalid node!")
```
